```python
import pandas as pd
import re
from scipy import stats

inv = '2Rb'

df = pd.read_csv(f"../old_results/karyotype/{inv}.BouakePM.karyo.txt", sep="\t")
df['inv_frequency'] = df[f'{inv} frequency'].astype(str).str.rstrip('%').astype(float) / 100

# Extract the group label (remove trailing digits from sampleID)
df["group"] = df["sampleID"].apply(lambda x: re.sub(r"\d+$", "", x))

df = df.query("group not in ['Kisumu',  'Ngousso']")

# --- Within-species comparisons ---
# Coluzzii: Cont vs PM
coluzzii_cont = df.loc[df["group"] == "coluzziiCont", "inv_frequency"]
coluzzii_pm   = df.loc[df["group"] == "coluzziiPM", "inv_frequency"]

t_col, p_col = stats.ttest_ind(coluzzii_cont, coluzzii_pm, equal_var=False)
print(f"Coluzzii: Cont vs PM -> t={t_col:.3f}, p={p_col:.3f}")

# Gambiae: Cont vs PM
gambiae_cont = df.loc[df["group"] == "gambiaeCont", "inv_frequency"]
gambiae_pm   = df.loc[df["group"] == "gambiaePM", "inv_frequency"]

t_gam, p_gam = stats.ttest_ind(gambiae_cont, gambiae_pm, equal_var=False)
print(f"Gambiae: Cont vs PM -> t={t_gam:.3f}, p={p_gam:.3f}")
```
