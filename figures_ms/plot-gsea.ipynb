```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import font_manager


# Color settings
golden_color = 'lemonchiffon'
header_color = 'lightsteelblue'
header_text_color = 'white'
golden_alpha = 0.3

# Font settings
font_path = '/home/sanj/.local/share/fonts/Diverda Sans W01 Regular.ttf'
font_prop = font_manager.FontProperties(fname=font_path)
plt.rcParams['font.family'] = font_prop.get_name()

df = pd.read_csv("../old_results/gsea/genediff/gambiaeCont_gambiaePM.de.tsv", sep="\t")

# Select only the columns we want
display_df = df[['annotation', 'descriptions', 'padj']].copy()

# Format padj for better display
display_df['padj'] = display_df['padj'].apply(lambda x: f"{x:.2e}" if x < 0.01 else f"{x:.3f}")

# Create the plot
fig, ax = plt.subplots(figsize=(14, 8))
ax.axis('tight')
ax.axis('off')

# Create table
table = ax.table(cellText=display_df.values,
                colLabels=['GO Term', 'Description', 'Adjusted P-value'],
                cellLoc='left',
                loc='center',
                colWidths=[0.15, 0.65, 0.2])

# Style the table
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1, 2)

# Color alternating rows golden
for i in range(len(display_df)):
    if i % 2 == 0:  # Even rows (0, 2, 4, ...)
        for j in range(len(display_df.columns)):
            table[(i+1, j)].set_facecolor(golden_color)
            table[(i+1, j)].set_alpha(golden_alpha)

# Style header
for j in range(len(display_df.columns)):
    table[(0, j)].set_facecolor(header_color)
    table[(0, j)].set_text_props(weight='bold', color=header_text_color)

plt.title('GSEA Results', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```
